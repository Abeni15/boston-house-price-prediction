```python
from sklearn.model_selection import train_test_split

y = df['MEDV']
# Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)
# Print the shapes of the training and testing feature sets to verify their dimensions
print(X_train.shape)
print(X_test.shape)
```

```python
import numpy as np

# Add a column of ones to X_train and X_test to serve as the intercept term for the linear regression model
X_train = np.c_[np.ones(X_train.shape[0]), X_train]
X_test = np.c_[np.ones(X_test.shape[0]), X_test]
```

```python
# Initialize the theta (coefficient) vector with zeros. Its size is determined by the number of features in X_train (including the intercept).
theta = np.zeros(X_train.shape[1])
```

```python
import matplotlib.pyplot as plt
import numpy as np

alpha = 0.01
iterations = 1000
m = len(y_train)
lambda_ = 0.1  # Define the regularization parameter (you can change this value)

# Initialize lists to store metrics for plotting
iterations_list = []
train_errors = []
val_errors = []

# Perform Gradient Descent to train the linear regression model with L2 regularization
for i in range(iterations):
    # Record iteration
    iterations_list.append(i)

    # Calculate predictions and errors for training set
    predictions_train = X_train @ theta
    errors_train = predictions_train - y_train

    # Calculate gradients with L2 regularization term
    gradients = (1/m) * (X_train.T @ errors_train) + (lambda_ / m) * theta
    theta = theta - alpha * gradients

    # Calculate and store training MSE
    train_mse = np.mean(errors_train ** 2)
    train_errors.append(train_mse)

    # Calculate predictions and errors for validation (test) set
    predictions_val = X_test @ theta
    errors_val = predictions_val - y_test

    # Calculate and store validation MSE
    val_mse = np.mean(errors_val ** 2)
    val_errors.append(val_mse)

```

```python
# Make predictions on the test set using the learned coefficients (theta)
y_pred = X_test @ theta
```

```python
# Calculate the Mean Squared Error (MSE) to evaluate the model's performance
mse = np.mean((y_test - y_pred) ** 2)
print("MSE:", mse)
```
